```python
import numpy as np
import matplotlib.pyplot as plt

# ---- Site & inputs ----
lat_deg = 28.76          # site latitude; change to yours
dec_deg = 22.76          # the declination track you want
zenith = 10.0            # ZA for your points (Alt = 80°)
az_points = [102.199, 248.117, 175.158, 355.158, 100.758, 259.265, 102.217,
             126.053, 275.258, 214.263, 248.099, 270, 90, 120.311, 230.005,
             262.712, 87.604]

# ---- Helpers ----
def altaz_from_dec_H(dec_deg, H_deg, lat_deg):
    phi = np.deg2rad(lat_deg)
    dec = np.deg2rad(dec_deg)
    H   = np.deg2rad(H_deg)
    # altitude
    sin_alt = np.sin(phi)*np.sin(dec) + np.cos(phi)*np.cos(dec)*np.cos(H)
    alt = np.arcsin(np.clip(sin_alt, -1, 1))
    # azimuth (0°=N, increasing to E)
    y = -np.sin(H)
    x = np.tan(dec)*np.cos(phi) - np.sin(phi)*np.cos(H)
    az = np.arctan2(y, x) % (2*np.pi)
    return az, alt

def hour_angles_for_alt(dec_deg, alt_deg, lat_deg):
    phi = np.deg2rad(lat_deg)
    dec = np.deg2rad(dec_deg)
    a   = np.deg2rad(alt_deg)
    denom = np.cos(phi)*np.cos(dec)
    if np.isclose(denom, 0.0):
        return np.array([])
    cosH = (np.sin(a) - np.sin(phi)*np.sin(dec)) / denom
    if (cosH < -1) or (cosH > 1):
        return np.array([])  # altitude not reachable
    H = np.arccos(np.clip(cosH, -1, 1))
    return np.array([-H, H])  # east/west solutions

# ---- Build declination track (sweep hour angle) ----
H_deg = np.linspace(-180, 180, 1441)
az_track, alt_track = altaz_from_dec_H(dec_deg, H_deg, lat_deg)
mask = alt_track > 0
theta_track = az_track[mask]           # polar angle
r_track = np.deg2rad(90 - np.rad2deg(alt_track[mask]))  # radius = ZA (radians)

# ---- Intersections of this declination with ZA=10° (Alt=80°) ----
altsol = 90 - zenith
Hs = hour_angles_for_alt(dec_deg, altsol, lat_deg)   # in radians
theta_cross = []
r_cross = []
for H in Hs:
    az_cross, alt_cross = altaz_from_dec_H(dec_deg, np.rad2deg(H), lat_deg)
    theta_cross.append(az_cross)
    r_cross.append(np.deg2rad(zenith))
theta_cross = np.array(theta_cross)
r_cross = np.array(r_cross)

# ---- Plot ----
fig = plt.figure(figsize=(7,7))
ax = plt.subplot(111, projection='polar')
ax.set_theta_zero_location('N')   # 0° at North
ax.set_theta_direction(-1)        # clockwise

# declination track
ax.plot(theta_track, r_track, lw=2, label=f"Declination δ = {dec_deg:.2f}°")

# your points at ZA=10
ax.scatter(np.deg2rad(az_points), np.deg2rad([zenith]*len(az_points)),
           s=40, label=f"Your points (ZA={zenith:.0f}°)")

# true intersections of this declination with ZA=10
if len(theta_cross):
    ax.scatter(theta_cross, r_cross, s=90, marker='x', lw=2,
               label="δ track ∩ ZA=10°")

# cosmetics
ax.set_rlim(0, np.deg2rad(90))
ax.set_rticks(np.deg2rad([10, 30, 50, 70, 90]))
ax.set_rlabel_position(45)
ax.set_title("Declination track δ=22.76° with ZA=10° points", va='bottom')
ax.grid(True)
ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1), fontsize=9)

plt.show()
```
