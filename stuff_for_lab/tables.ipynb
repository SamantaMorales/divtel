```python
class Friend:
    def __init__(self, name):
        self.name = name
        self.paid = 0.0

    def add_payment(self, amount):
        self.paid += amount

    def __repr__(self):
        return f"Friend(name={self.name!r}, paid={self.paid:.2f})"

alice = Friend("Alice")
alice.add_payment(20)
print(alice)  
```
